```python
import numpy as np
list1=np.array([[1,2,3],[23,4,5]])
print(list1)
print(type(list1))
list2=np.array([[[1,2,3],
                 [8,9,7],
                 [5,6,7]]])
print(list2)
```

```python
a=np.array([10,20,30])
print(a.dtype)
```

```python
a=np.array([10,20,30])
print(a.ndim)
```

```python
a=np.array([[1,2,3],[23,4,5],[4,5,6]])
print(a)
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

```python
zero=np.zeros(5,dtype=int)
print(zero)
```

```python
a=np.ones(5,dtype=int)
print(a)
```

```python
a=np.eye(6,dtype=int)
print(a)
```

```python
arr=np.arange(0,16,2)
print(arr)
```

```python
a=np.linspace(1,100,6,dtype=int)
print(a)
```

```python
a=np.arange(1,13)
print(a.reshape(3,4))
```

```python
a=np.arange(1,31)
print(a.reshape(5,2,3))
```

```python
b=np.array([[[1,2,3],[7,8,6],[9,0,6]]])
print(b.flatten())
```

```python
a=np.array([1,2,3])
print(a*10)
```

```python
a=np.array([1,2,3])
print(a.mean())
```

```python
a=np.array([[1,2,3],[4,5,6]])
print(a.sum(axis=0))
print(a.sum(axis=1))
```

```python
a=np.array([1,2,3])
print(a.std())
```

```python
a=np.array([1,2,3])
print(a.var())
```
